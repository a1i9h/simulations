Code for Silicon with confinement constant 10

```python
#E(r)=E(bulk)+A/r^2
#wavelength = hc/E

import numpy as np
from matplotlib import pyplot as plt

r = np.linspace(1, 10, 10)  # radius of QD
A=10
Eb=1.1
E = Eb + A / r**2
w=1240/E
plt.plot(r,w,color='b')
plt.title('Wavelength vs QD Size')
plt.show()
```
