```python
#std_dev= (4kTRΔf)^0.5

import numpy as np
from matplotlib import pyplot as plt
N=int(input("enter a value:"))
T=int(input("enter temperature:"))
R=int(input("enter resistance:"))
f=int(input("enter bandwidth:"))
k=1.380649*10**-23
std_dev= (4*k*T*R*f)**0.5
noise = np.random.normal(0, std_dev, N)
plt.hist(noise, bins=50)
plt.show()
```
