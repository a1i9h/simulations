```python
#E(r)=E(bulk)+A/r^2
import numpy as np
from matplotlib import pyplot as plt

r = np.linspace(1, 10, 10) #radius of QD
ele = int(input("Enter 1 for Si, 2 for CdSe: "))

if ele == 1:
    Eb = 1.1  
else:
    Eb = 1.7 

A = 10
E = Eb + A / r**2
plt.plot(r, E, color='r', label='Band gap variation')
plt.title('Band Gap vs QD Size')
plt.show()
```

Variation of band gap with A

```python
import numpy as np
from matplotlib import pyplot as plt

r = np.linspace(1, 10, 10)  
ele = int(input("Enter 1 for Si, 2 for CdSe: "))

if ele == 1:
    Eb = 1.1  # Silicon
else:
    Eb = 1.7  # CdSe

A_values = np.linspace(1, 10, 5)  

for A in A_values:
    E = Eb + A / r**2
    plt.plot(r, E, label=f'A = {A:.2f}') 


plt.title('Band Gap vs QD Size for Different A')
plt.legend()
plt.show()
```
