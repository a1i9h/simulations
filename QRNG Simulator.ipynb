```python
import numpy as np
import matplotlib.pyplot as plt

noise = np.random.normal(0, 1e-9, 800)
bits = (noise > 0).astype(int)

bits = bits[:len(bits) - len(bits) % 8]
nums = [int("".join(map(str, bits[i:i+8])), 2) for i in range(0, len(bits), 8)]

plt.hist(bits, bins=[-0.5, 0.5, 1.5])  
plt.xticks([0, 1])
plt.title("Bit Distribution from Simulated Noise")
plt.show
print("Random numbers:", nums[:10])
```
